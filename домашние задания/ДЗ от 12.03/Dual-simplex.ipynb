```python
from itertools import product


def build_hamming_parity_check_matrix(m: int):
    n = 2**m - 1
    columns = []

    for x in range(1, 2**m):
        col = [(x >> i) & 1 for i in range(m)]
        columns.append(col)

    H = [[columns[j][i] for j in range(n)] for i in range(m)]
    return H


def xor_vectors(a, b):
    return [(x ^ y) for x, y in zip(a, b)]


def vector_to_str(v):
    return ''.join(map(str, v))


def weight(v):
    return sum(v)


def generate_dual_codewords(H):
    m = len(H)
    n = len(H[0])
    result = []

    for coeffs in product([0, 1], repeat=m):
        word = [0] * n
        for c, row in zip(coeffs, H):
            if c == 1:
                word = xor_vectors(word, row)
        result.append((coeffs, word))

    return result


def save_and_print_result(m: int, filename: str = None):
    if filename is None:
        filename = f"dual_hamming_m{m}.txt"

    H = build_hamming_parity_check_matrix(m)
    codewords = generate_dual_codewords(H)

    n = 2**m - 1
    expected_weight = 2**(m - 1)

    lines = []
    lines.append(f"ПРОВЕРКА: дуальный код к коду Хэмминга при m = {m}")
    lines.append(f"Длина кода n = 2^{m} - 1 = {n}")
    lines.append(f"Ожидаемый ненулевой вес для симплекс-кода: {expected_weight}")
    lines.append("")
    lines.append("Проверочная матрица H:")
    for row in H:
        lines.append("  " + vector_to_str(row))
    lines.append("")
    lines.append("Кодовые слова дуального кода:")
    lines.append("")

    nonzero_weights = set()

    for idx, (coeffs, word) in enumerate(codewords, start=1):
        w = weight(word)
        if w != 0:
            nonzero_weights.add(w)

        coeffs_str = vector_to_str(coeffs)
        word_str = vector_to_str(word)

        lines.append(
            f"{idx:>2}) коэффициенты: {coeffs_str}   слово: {word_str}   вес: {w}"
        )

    lines.append("")
    lines.append("Итоги:")
    lines.append(f"  Всего слов: {len(codewords)}")
    lines.append(f"  Уникальные ненулевые веса: {sorted(nonzero_weights)}")

    if nonzero_weights == {expected_weight}:
        lines.append(
            f"  Вывод: все ненулевые кодовые слова имеют вес {expected_weight}."
        )
        lines.append(
            "  Следовательно, дуальный код к коду Хэмминга действительно является симплекс-кодом."
        )
    else:
        lines.append("  Вывод: свойство симплекс-кода не выполнено.")

    text = '\n'.join(lines)

    print(text)

    with open(filename, "w", encoding="utf-8") as f:
        f.write(text)



if __name__ == "__main__":
    save_and_print_result(m=3)
```
